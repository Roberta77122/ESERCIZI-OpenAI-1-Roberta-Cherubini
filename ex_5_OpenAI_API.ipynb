```python

# TODO Inizializza il client di Open AI

import os
import openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')

openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python
input_text = input("Enter text: ")


completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": 
            """
            Leggi il testo e analizzalo. 
            Restituisci cinque domande inerenti il testo e volte a determinare il livello di comprensione
            dell'utente.
            """
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
)

print(completion.choices[0].message.content)
```
