**Traccia:** Crea un prompt che riassuma un articolo di notizie in un singolo paragrafo di al massimo 255 caratteri.

**Obiettivo:** Prendere un testo lungo e condensarlo nei punti essenziali.

```python

# TODO Inizializza il client di Open AI

import os
import openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')

openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()


```

```python
input_text = input("Inserisci un articolo: ")

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
                    "role": "system",
                    "content": """
                        Leggi attentamente l'articolo e 
                        riassumilo in meno di 256 caratteri.
                        Cerca di trattare tutti gli argomenti più importanti/essenziali
                        senza superare il limite suddetto.
                        """
                },
                {
                    "role": "user",
                    "content": input_text
        
                }
    ],
)

print(completion.choices[0].message.content)
```
